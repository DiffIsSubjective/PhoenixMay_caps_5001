6. (2 pts) You need to create a copy of an important list and alter the copy. Why is the following code a bad idea? How do you fix it so its no longer a bad idea?

```python
list1 = [2,4,6]
list2 = list1
list2[0]= 8
```

```python
#This code is a poor choice because it's not really making a copy. list1 and list2 are referencing the exact same point in memory, so changing list2 will also
#change list1. This defeats the purpose of making a copy you can edit. Instead, let's use the copy() function:

#Establishes the first list, then uses copy() to make a copy of it that is NOT stored in the same memory location.
list1 = [2,4,6]
list2 = list1.copy()

#Then we make a change to the first item in list2
list2[0] = 8

#Then we print both lists to prove we did not alter both:
print(list1)
print(list2)
```

7. (2 pts) What is truthy and falsey? Provide a list example **and** a string example of something that is truthy and something that is falsey.

```python
#Truthy and falsy have to do with boolean evaluations. Something "truthy" will have a boolean evaluation of True, and something falsy will have a
#boolean evaluation of False. See some examples:

#List example of truthy:
L = [1,2,3]
if L:
    print('This prints because L is truthy.')

#List example of falsy:
M = []
if M:
    print('This does not print because M is falsy.')

#String example of truthy:
A = "Nickelback is good and Im tired of pretending its not."
if A:
    print('This prints because A is truthy.')

#String example of falsy:
B = ""
if B:
    print('This will not print because B is falsy.')
```

12. Provide an expression that evaluates which variable is greater. 

```python
foo = 77
bar = 38

#This expression uses simple subtraction to obtain the difference between variables. If it's positive, the first variable (foo) is greater,
#and if it's negative, the second variable (bar) is greater. If it's zero then they are equal.
expression = foo - bar
if expression > 0:
    print("foo is greater than bar.")
if expression < 0:
    print("bar is greater than foo.")
if expression == 0:
    print("foo and bar are equal.")
```

17. Slice `some_string` to only include the words `quarter to eight`. 

```python
some_string = "I have got a date at a quarter to eight; I'll see you at the gate, so don't be late"
```

```python
#This establishes the original some_string text string, finds where the "quarter to eight text" starts and how long it is, then slices the string to only
#include that part. Then we print it to make sure it worked.

some_string = "I have got a date at a quarter to eight; I'll see you at the gate, so don't be late"

place = some_string.find("quarter to eight")
length = len("quarter to eight")
some_string = some_string[place:place+length]

print(some_string)
```
