```python
"""
This just creates a scatter plot of the Rate of New Marijuana Users per
1000 people and the Rate of All Marijuana Users per 1000. The plot
seems to show a pretty strong 
"""

import pandas as pd
import matplotlib.pyplot as plt

drugs = pd.read_csv("drugs.csv")

plt.plot(drugs["Rates.Marijuana.New Users.12-17"], drugs["Rates.Marijuana.Used Past Month.12-17"], ".")
plt.xlabel("Rate of New Marijuana Users per 1000")
plt.ylabel("Rate of All Marijuana Users for Past Month per 1000")
plt.title("Scatter Plot of New vs All Rates of Marijuana Usage for Age 12-17")
plt.show()
```
