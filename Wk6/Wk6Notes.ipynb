```python
# Example 4: raising an exception yourself with the raise keyword. This triggers an exception
# through the code and specifies one of the default types.

try:
    raise IndexError
except IndexError:
    print("This is an index error")
```

```python
# Ex 9: except without exception type (not encouraged, since it's hard to diagnose what occurred)
try:
    z = 1/0
except:
    print("Some kind of exception happened!")
```

```python
# Ex 10: except with "Exception" object type
try:
    z = 1/0
except Exception:
    print("Some kind of exception happened!")
```

```python
# Ex 11: repr function
try:
    x = int("not a number")
except Exception as e:
    print(f"Caught exception: {e}.")
    print(f"Caught exception: {repr(e)}.")
```

```python
class a_sample_class:

    #default/basic constructor
    def __init__(self):
        self.var = 0
    def print_vars(self):
        print(self.var)

obj = a_sample_class()

obj.print_vars()
```

```python
class a_sample_class:
    class_attr = 5
    #parameterized constructor
    def __init__(self, one, two):
        self.first = one
        self.second = two
    #an instance method for printing
    def print_vars(self):
        print(self.first)
        print(self.second)

#create instance of class
obj = a_sample_class(17,42)
#call instance method to get values
#for inserted parameters
obj.print_vars()
```
