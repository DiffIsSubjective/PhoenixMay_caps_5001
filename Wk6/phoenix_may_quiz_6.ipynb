```python
"""
Question 1: Using the raise statement to raise a ZeroDivisionError.
"""
try:
    a = 10 / 0
except ZeroDivisionError as exc:
    raise ZeroDivisionError ("Cannot divide by zero!") from exc
else:
    print(f"The result is {a}.")

# If the "try" division problem divides by zero, then the "else" does not execute.
# But if it does not try to divide by zero, the "else" will execute.
```

## Question 3: Recommended Approach

Why is the approach below a recommended approach to exception handling over the generic except?

except Exception as e:

...

This is recommended for two reasons:

1. Because it allows you to return the Exception in your code's output. If you just use the generic "except," your code is not storing the exception anywhere, so you won't be able to say what it was in your output. But if you use the "except Exception as e" line, you can put the variable e in your printed output so the exact exception is shown to the user.

2. Because just using "except" might catch a genuine programming error. This is related to the first item above, but adding in the possibility of an error in the program itself will make the actual issue much harder to diagnose.
