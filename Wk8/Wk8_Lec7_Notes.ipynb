```python
# Warm up exercise for 10/20/2026

def fizz_buzz(num: int):
    """
    This function plays a variable number of fizzbuzz rounds.

    num = number that the user wants to play. Should be an integer.

    The function loops through every number from 1 to num (inclusive) and prints "FizzBuzz" if the number
    is divisible by 3 and 5, "Fizz" if only divisible by 3, and "Buzz" if only divisible by 5. If it's
    not divisible by either, it prints the number.
    """
    for i in range(1,num+1):
        if i % 3 == 0 and i % 5 == 0:
            print("FizzBuzz")
        elif i % 3 == 0:
            print("Fizz")
        elif i % 5 == 0:
            print("Buzz")
        else:
            print(i)

fizz_buzz(30)

```
