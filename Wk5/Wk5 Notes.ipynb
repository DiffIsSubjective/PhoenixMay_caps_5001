```python
import time

L = []
start_time = time.time()
for x in range (10):
    L.append(x+10)
print(L)
end_time = time.time()
print(f"{end_time - start_time} seconds elapsed.")

# using list comprehension
start_time = time.time()
L = [x+10 for x in range (10)]
print(L)
end_time = time.time()
print(f"{end_time - start_time} seconds elapsed")
```

```python
def add_numbers(a: int | float, b: int | float) -> int | float: # We should have doc strings for functions, as seen below.
    """
    Adds two numbers together.

    Args:
    a (int or float): the first number
    b (int or float): the second number

    Returns:
    int or float: the sum of both numbers.
    """
    y = a + b
    return y

add_numbers(5,6)
```

```python
from datetime import date, datetime

def parse_date(value):
    #Return a datetime.date or None if the date is invalid.
    if isinstance(value, date):
        return value

    if not isinstance(value, str):
        return None

    for fmt in ["%m-%d-%Y", "%m/%d/%Y", "%Y-%m-%d", "%Y/%m/%d"]:
        try:
            return datetime.strptime(value, fmt).date()
        except ValueError:
            continue

    return None

type(parse_date("2028/2/29"))
```
