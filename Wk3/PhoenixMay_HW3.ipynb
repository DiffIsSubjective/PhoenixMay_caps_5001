1. (3 pts) The string "a\nb\x1f\000d" contains the bytes a, newline (\n), b, binary 31 (a
hex escape \x1f), binary 0 (an octal escape \000), and d. 
- Pass the string to the built-
in len function to verify this, and print each of its character’s ord results to see the
actual byte values. See Table 7-2 for more details.
- Convert the string to a human readable format

```python
string = "a\nb\x1f\000d"

# This prints the string length of 6
string_length = len(string)
print(f"The total length of the string is {string_length}.")

# This prints the ord results to show the byte values
ord_list = [ord(i) for i in string]
print(f"The ord values for each character in the string are as follows:{ord_list}.")

# This makes a dictionary with keys of the byte values we want to replace, and
# values of what I want to replace it with. I tried using an ordered list at first,
# but that wasn't working how I wanted. I asked Gen AI about using lists for values to
# replace in a string and it suggested a dictionary instead, which worked better.
replacements = {0: "[0]", 10: "[NEWLINE]", 31: "[31]"}

# This iterates thru the string and replaces things based on the above replacements dictionary.
# This works by establishing char_conversions as a blank list, checking if a character is
# in the replacements dictionary, adding the replacement text if it is, and adding the original
# character if it isn't.
char_conversions = [""]
for char in string:
    byte = ord(char)
    if byte in replacements:
        char_conversions.append(replacements[byte])
    else:
        char_conversions.append(char)

# This turns the char_conversions list into a string. The blank quotes are because I'm not
# using any special characters between the values.
new_string = "".join(char_conversions)

# Prints the human-readable string
print(f"The human-readable version of the string is {new_string}.")
    
```

```python
data = {'state':['Ohio','Ohio','Ohio','Nevada','Nevada','Nevada'],

'year':[2000,2001,2002,2001,2002,2003],

'pop':[1.5, 1.7, 3.6, 2.4, 2.9, 3.2]}

# Prints the number of keys in the dictionary.
print(f"There are {len(data.keys())} keys in the dictionary.")

# Checks if the "pop" key exists. Since the 'pop in data' statement is truthy if it's a key, but falsy if
# it isn't, this works. I checked it by changing the "pop" key to another name and it output the "Key not found".
if 'pop' in data:
    print("Pop key is in dictionary.")
else:
    print("Key not found.")

# Adds the year 3000 to the "year" key. Gen AI was used to find out I could use empty brackets to reference the whole list.
data.setdefault('year', []).append(3000)

# Uses the pop method to remove the state key and immediately assigns it to the "states" variable.
# Then we create a new dictionary using the states key.
states = data.pop('state')
state_data = {'state': states}

# Sorts the new dictionary of states alphabetically by default.
state_data['state'].sort()

# Loops thru data and prints key name and length.
for key in data:
    print(f"The key {key} has a length of {len(data[key])}.")

# This brings up the pop key and slices the list to remove its first element.
# Gen AI was used to determine how I could reference specific items within a list within a dictionary.
# I didn't realize you could use two pairs of [][] next to each other to reference the key and index.
data['pop'] = data['pop'][slice(1,len(data['pop']))]

# Adds a new key called 'region' to the data dictionary.
data['region'] = ['Midwest', 'Midwest', 'Midwest','West','West','West']

# This prints the final versions of both dictionaries to ensure everything worked.
# The 'year' key in data has the added 3000, the 'state' key is no longer in data, the state-data dictionary is sorted
# alphabetically, the 'pop' key in data is missing its original first element (1.5), and data now has the 'region' key.
print(data)
print(state_data)
```

states = [' Alabama ', 'Georgia!', 'Georgia', 'georgia',
'FlOrIda', 'south carolina##', 'West virginia?’]

```python
# The list in the above markdown cell ends in a ` instead of a ' which causes an unterminated string error.
# I replaced it with ' so the list works. Other than that, I did not edit the original list.
states = [' Alabama ', 'Georgia!', 'Georgia', 'georgia','FlOrIda', 'south carolina##','West virginia?']

# The first line strips the extra spaces & special characters out of the state names.
strip_states = [state.strip(',#,!,?') for state in states]
# This next line converts everything to initial capital letters.
norm_states = [state.title() for state in strip_states]

print(norm_states)
```

5. (2 pts) Convert the original dictionary in question 2 to a DataFrame. 
- Select only the pop column and get the mean of the values in that column. 

```python
import pandas as pd

# Since question 2 had us alter the list, this section re-establishes the original version of the list as original_data:
original_data = {'state':['Ohio','Ohio','Ohio','Nevada','Nevada','Nevada'],

'year':[2000,2001,2002,2001,2002,2003],

'pop':[1.5, 1.7, 3.6, 2.4, 2.9, 3.2]}

# This line creates the DataFrame using pandas.
df_data = pd.DataFrame(original_data)

# This line creates a list containing only the pop column.
pop_col = df_data["pop"]
# This line calculates the mean of the pop column.
pop_mean = sum(pop_col) / len(pop_col)

print(df_data)
# This rounds the mean to 2 decimal places and prints it.
print(f"The mean of the pop values is {pop_mean:.2f}.")
```

6. (2 pts) Loop through the string and eliminate all the asterisks.

```python
asterisk = 'My* compu*ter has go*ne craz***y an*d add*ed extra *symbols.*'

# This sets up no_asterisk as a blank string at first. Without this line, no_asterisk contained the original
# string *and* the version without asterisks. Gen AI was used to come up with the idea to establish it as
# a blank string before the loop. I was curious, so I asked Gen AI why I had that strange error in the first place:
# Apparently, I had this unique error because the *first* time I used that variable name, my code didn't work,
# so Python had already assigned it the original string's value. When I re-tried with this loop, Python started
# with that original string and then added the asterisk-less one onto the end. I experimented by removing the
# below line and using a new variable name, and sure enough, Python gave me a "variable not defined" error.
# I never considered Python using variable values from old versions of scripts that have been changed. Neat!
no_asterisk = ""

# This loops through the string, checks if a character is an asterisk, and if it's not, it is added to the
# no_asterisk string.
for ltr in asterisk:
    if ltr not in "*":
        no_asterisk += ltr

print(no_asterisk)
```
