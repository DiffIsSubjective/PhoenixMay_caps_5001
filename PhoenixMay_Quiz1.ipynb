Strings, numbers, and tuples are immutable. Strings are sequences of characters, numbers are sequences of digits, and tuples are sequences of different values (like a list, but immutable). See the next cell for a code example of tuples being immutable:

```python
#Question 5 cont'd

#This establishes a tuple with the values 1, 2, 3, and 4.
T = (1,2,3,4)

#This attempts to change the first value of the tuple from 1 to 99.
T[0] = 99

#This attempts to print the tuple. It will result in an error, since tuples
#are immutable and we attempted to change it anyway.
print(T)
```
