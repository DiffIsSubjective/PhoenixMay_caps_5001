3. (2 pts) Write code to determine which car has the highest MPG, lowest MPG, and return other relevant information about it. 

```python
# Python kept splitting my output into multiple rows of a few columns each, so I used Gen AI to come up with this line to make the output wider.
pd.set_option('display.width', 1000)

# This creates a dataframe with ; as the delimiter, since Python interprets each row as having one value otherwise.
# Used Gen AI to determine how to clarify the separator as ;
cars_df = pd.DataFrame(pd.read_csv(cars_pd, sep=';'))

# This searches for the DF rows with the highest MPG values and prints all of them.
print("The car(s) with the highest MPG:\n")
max_mpg = cars_df[cars_df['MPG'] == cars_df['MPG'].max()]
print(max_mpg)

# This searches for the DF rows with the lowest MPG values and prints all of them.
print("\nAnd the car(s) with the lowest MPG:\n")
min_mpg = cars_df[cars_df['MPG'] == cars_df['MPG'].min()]
print(min_mpg)
```

4. (3 pts) Create a new pandas dataframe that has the average mpg, weight, and horsepower per country of origin stored as a tuple in that column. What does this tell us? Why might you use a tuple? Be mindful of how you name the columns so information isnt lost. 

```python
# This creates a new DataFrame that groups the original one by country of origin.
# I know Europe is a continent, not a country, but I'm going to refer to all of these as countries for simplicity.
country_cars_df = cars_df.groupby('Origin')

# This calculates the average MPG, weight, and horsepower for each country in the above DataFrame and prints it.
# This is just so I can easily check that the DataFrame with tuples, created in the next section, has the right results.
averages = country_cars_df[["MPG", "Weight", "Horsepower"]].mean()
print(f"DataFrame that does not use tuples:\n{averages}\n")

# This establishes a list for each country of origin. The list contains the values from the above DataFrame converted to floats.
# If I don't convert them to floats, there's a bunch of extra text that appears before each number.
eur = [float(averages.loc['Europe', 'MPG']), float(averages.loc['Europe', 'Weight']), float(averages.loc['Europe', 'Horsepower'])]
jap = [float(averages.loc['Japan', 'MPG']), float(averages.loc['Japan', 'Weight']), float(averages.loc['Japan', 'Horsepower'])]
usa = [float(averages.loc['US', 'MPG']), float(averages.loc['US', 'Weight']), float(averages.loc['US', 'Horsepower'])]

# This establishes three blank lists so I can add to them in the next section.
eur_rd = []
jap_rd = []
usa_rd = []

# This iterates through the indices of each list created earlier and rounds the float values to 2 decimal places. It then
# adds the rounded values to the three blank lists created above.
for x in range(0,3):
    eur_rd.append(round(eur[x],2))
    jap_rd.append(round(jap[x],2))
    usa_rd.append(round(usa[x],2))

# This converts the lists of each country's averages into tuples. You can hover your mouse over the variable names to show
# that they are tuples.
eur_rd = tuple(eur_rd)
jap_rd = tuple(jap_rd)
usa_rd = tuple(usa_rd)

# This creates a dictionary with the three countries and the tuples of averages we made above. This ensures that
# the tuples will be shown in one column per the instructions.
tuple_cars = {"Origin": ["Europe", "Japan", "US"],
              "Average MPG, Weight, & Horsepower": [eur_rd, jap_rd, usa_rd]}

# This converts the above dictionary into a DataFrame and then prints it.
tuple_df = pd.DataFrame(tuple_cars)
print(f"DataFrame that DOES use tuples:\n{tuple_df}")

#=================================== Answering the other questions ========================================================

# This tells us that the US has the lowest average MPG but the highest weight and horsepower, while Europe and Japan have pretty
# similar values across the board. Aggregating data into a few categories like this can make the dataset easier to digest, but
# it can also cut out a lot of the nuance. Strategies like this can tell you a lot by showing you little.

# You might use a tuple because of data integrity. Since the data cars.csv file is already recorded and won't be updated during this analysis,
# there's no need to store its values in a mutable data type. In this case, we have aggregations of its values in the form of averages
# by country of origin, but the same logic applies. This protects us from accidentally changing the values after calculating them.
```

```python
# PART 1
# This sets up the data structure as a library. The first four are some of my favorites, and the other four are ones where I thought it would
# be funny if they weren't checked out. No shade to Twilight though, I liked the movies.
library = {"Title": ["Rainbow Six", "The Road", "It", "Skeleton Crew", "Ulysses", "War and Peace", "Twilight", "Fifty Shades of Grey"],
           "Author": ["Tom Clancy", "Cormac McCarthy", "Stephen King", "Stephen King", "James Joyce", "Leo Tolstoy", "Stephenie Meyer", "E. L. James"],
           "Subject": ["Spy & war", "Post-apocalypse", "Horror", "Horror", "Modernism", "Historical fiction", "Romance", "Romance"],
           "Publication Year": ["1998", "2006", "1986", "1985", "1922", "1869", "2005", "2011"],
           "Status": ["On-loan", "On-loan", "On-loan", "On-loan", "In-library", "In-library", "In-library", "In-library"]}

# PART 2
# Some data structures do make more sense than others. Since the "status" is going to be changed when each book is checked out or returned,
# you want something mutable, so tuples are out (although you could make a mutable data structure within a tuple, like a list, to account for that).
# You also want something that can be organized by key or category, so one big ordered list would be too much of a pain to work with. I went with a
# dictionary first since it can be organized by key, and we've had exercises with transforming dictionaries into DataFrames, so I'm comfortable
# working with them. But I think a dictionary is better-suited at first since I don't see the library needing to use this for analysis most of the time.
# Although they could analyze data collected over time regarding which books are checked out more...but I digress.

# PART 3

# This sets up blank lists for the for-loop that comes next.
title_list = []
auth_list = []
sub_list = []
year_list = []
status_list = []

# This uses indexing to loop through the key values and appends them to lists for each key in library.

for x in library["Title"]:
    title_list.append(x)
for x in library["Author"]:
    auth_list.append(x)
for x in library["Subject"]:
    sub_list.append(x)
for x in library["Publication Year"]:
    year_list.append(x)
for x in library["Status"]:
    status_list.append(x)

# This prints the values of the lists created above on different lines. This shows that the indexing worked.

print(f"The titles of the books are: {title_list}.")
print(f"The authors of the books are: {auth_list}.")
print(f"The subjects of the books are: {sub_list}.")
print(f"The publication years of the books are: {year_list}.")
print(f"The status of the books are: {status_list}.")

# PART 4

import csv

# This creates the csv file, opens it in write mode, and establishes the column names.
with open('books.csv', 'w', newline = '') as csvfile:
    fieldnames = ['Title', 'Author', 'Subject', 'Publication Year', 'Status']
    writer = csv.DictWriter(csvfile, fieldnames = fieldnames)

# This loops through the index of each key in the dictionary and writes each row as a field with each book's info.
# X is set to step at the length of the "Title" key list in the dictionary, since that will contain all book titles. It works
# as a stopping point because the index starts at 0, so the 8th & final book is index [7], and then Python stops once x reaches 8.
    writer.writeheader()
    for x in range(0, len(library["Title"])):
        writer.writerow({'Title': library["Title"][x], 'Author': library["Author"][x], 'Subject': library["Subject"][x],
                         'Publication Year': library["Publication Year"][x], 'Status': library["Status"][x]})

# I checked the csv file in Excel and it worked. Each book has its own row, and the first row is headers.

# PART 5

# This establishes blank lists for the next part.
big_list = []
title = []
author = []
subject = []
pub = []
status = []

# This opens the file in read mode and takes each row, strips the /n, splits it by comma, and adds it to big_list.
# This way, big_list is a list of lists, and each sub-list is a row. It includes the headers in index[0].
# I could make this easier by using reader, but the instructions mentioned the split() function, so I wanted to make
# sure I utilized it in some way. If I used DictReader, it would be even easier, but that doesn't require the split() function.
with open('books.csv', 'r') as file:
    for line in file:
            big_list.append(line.strip().split(','))

# This goes through the indices of big_list and makes separate lists for each column name. It skips index 0 because
# those are all headings.
# I also manually input the index for the sub-list values, since the [0] index will always be the title, et cetera.
for x in range(1,len(big_list)):
   title.append(big_list[x][0])
   author.append(big_list[x][1])
   subject.append(big_list[x][2])
   pub.append(big_list[x][3])
   status.append(big_list[x][4])

# This prints all the lists above. One list per book component, with each list having the values for all books.
print("\nHere is the same info, but taken from the CSV and converted back into lists:")
print(f"Titles from CSV: {title}.")
print(f"Authors from CSV: {author}.")
print(f"Subjects from CSV: {subject}.")
print(f"Publication years from CSV: {pub}.")
print(f"Current status from CSV: {status}.")

# PART 6

# This makes a list of all index values of "In-library" in the staus list. Gen AI was used to assist with this, mainly
# for teaching me about the enumerate function.
available = [index for index, value in enumerate(status) if value == 'In-library']

# This iterates through the values in the available list and prints values with those indices from the title list.
print("\nHere is a list of titles currently available in the library:")
for x in available:
    print(title[x])
```
