The question presented in class was, "Is it intentional that a single-item Tuple in Python will be interpreted as the object type of the item, not as a Tuple?"

According to a Stackoverflow discussion I found (linked below), the answer is that it's unintentional. Since parenthesis are used quite a bit in expressions for grouping sub-sections, Python's creator decided it would be too confusing to make parenthesis mean a grouping *and* a tuple. So, tuples are not defined by the usage of parenthesis, but by the usage of commas within parenthesis. This is why (1) is assigned an object type of int, but (1,) is assigned the object type of tuple. See the below cell for a demonstration of this.

Stackoverflow link: https://stackoverflow.com/questions/40710455/why-single-element-tuple-is-interpreted-as-that-element-in-python

```python
not_tuple = (1)

print(f"The variable not_tuple has a type of {type(not_tuple)}.")

tuple = (1,)

print(f"The variable tuple has a type of {type(tuple)}.")
```
